```python
from graphviz import Digraph

dot = Digraph(comment='Information System Flow')

dot.attr('node', shape='box', style='filled', fontname='Arial', color='lightblue')
dot.node('A', '1. User fills form / enters data\n(user input interface)')
dot.node('B', '2. Raw data captured by the system\n(input processing)')
dot.node('C', '3. System checks: Is it TPS, MIS, etc?\n(categorization module)')
dot.node('D', '4. Saves into correct category\n(TPS, MIS, HRIS...)')
dot.node('E', '5. Summarizes the stored data\n(report generator)')
dot.node('F', '6. Report shown in table format\n(output display)')

dot.attr('node', color='lightgoldenrod1')
dot.node('V', 'Validation Check:\nCheck fields, types, missing values')

dot.edge('A', 'B')
dot.edge('B', 'C')
dot.edge('C', 'D')
dot.edge('D', 'E')
dot.edge('E', 'F')
dot.edge('V', 'B')

dot.render('info_system_flow', format='png', cleanup=False)
dot
```
